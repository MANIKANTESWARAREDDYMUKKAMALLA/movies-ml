# RQ7 — Decision matrix (qualitative comparison)

**Research question (RQ7).** Beyond raw predictive performance, how do candidate models compare across practical criteria (interpretability, training cost, robustness, deployment simplicity)?

**Task:** qualitative comparison (no fitting required). **Outputs:** CSV + PDF in `./outputs`.

## Methodology (this notebook)

- Build a small decision matrix with 0–5 scores.
- Scores are analyst-assigned (not learned).
- Save a table and radar-style chart.

```python
from __future__ import annotations

import os
from pathlib import Path

import matplotlib

matplotlib.use("Agg")
import matplotlib.pyplot as plt
import numpy as np
import pandas as pd

IS_KAGGLE = os.path.exists("/kaggle/input")
OUT = Path("/kaggle/working") if IS_KAGGLE else Path("outputs")
OUT.mkdir(parents=True, exist_ok=True)

RQ_PREFIX = "RQ07"

# 0–5 where higher is better
criteria = [
    "PredictivePower",
    "Interpretability",
    "TrainingCost",
    "Robustness",
    "DeploymentSimplicity",
]

# Adjust these if your RQ2/RQ5 results suggest different winners.
rows = [
    {
        "Model": "LinearRegression",
        "PredictivePower": 2,
        "Interpretability": 5,
        "TrainingCost": 5,
        "Robustness": 3,
        "DeploymentSimplicity": 5,
    },
    {
        "Model": "Ridge",
        "PredictivePower": 3,
        "Interpretability": 4,
        "TrainingCost": 5,
        "Robustness": 4,
        "DeploymentSimplicity": 5,
    },
    {
        "Model": "RandomForest",
        "PredictivePower": 4,
        "Interpretability": 2,
        "TrainingCost": 3,
        "Robustness": 4,
        "DeploymentSimplicity": 3,
    },
    {
        "Model": "GradientBoosting",
        "PredictivePower": 4,
        "Interpretability": 2,
        "TrainingCost": 3,
        "Robustness": 4,
        "DeploymentSimplicity": 3,
    },
    {
        "Model": "XGBoost",
        "PredictivePower": 5,
        "Interpretability": 2,
        "TrainingCost": 2,
        "Robustness": 4,
        "DeploymentSimplicity": 3,
    },
]

tbl = pd.DataFrame(rows)
tbl["Total"] = tbl[criteria].sum(axis=1)
tbl = tbl.sort_values(["Total", "PredictivePower"], ascending=False)

tbl.to_csv(OUT / f"{RQ_PREFIX}_table_decision_matrix.csv", index=False)

# Radar chart
labels = criteria
angles = np.linspace(0, 2 * np.pi, len(labels), endpoint=False).tolist()
angles += angles[:1]

fig = plt.figure(figsize=(7.5, 6))
ax = plt.subplot(111, polar=True)

for _, r in tbl.iterrows():
    vals = [r[c] for c in labels]
    vals += vals[:1]
    ax.plot(angles, vals, linewidth=2, label=r["Model"])
    ax.fill(angles, vals, alpha=0.07)

ax.set_xticks(angles[:-1])
ax.set_xticklabels(labels)
ax.set_yticks([1, 2, 3, 4, 5])
ax.set_yticklabels(["1", "2", "3", "4", "5"])
ax.set_title("RQ7 — Decision matrix radar (qualitative scores)")
ax.legend(loc="upper right", bbox_to_anchor=(1.35, 1.15))

plt.tight_layout()
fig.savefig(OUT / f"{RQ_PREFIX}_fig_decision_matrix_radar.pdf")
plt.close()

tbl
```
